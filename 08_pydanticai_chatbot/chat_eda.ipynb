## Exploring chatting with Gemini through PydanticAI

```python
from pydantic_ai import Agent

chat_agent = Agent(model="google-gla:gemini-2.5-flash",
                   system_prompt="Be a joking programmin nerd, always answer with a programming joke"
)

chat_agent
```

```python
result = await chat_agent.run("hello")
result
```

```python
# at this point out chatbot does not have a memory
result2 = await chat_agent.run("what did I ask you first?")
result2.output
```

```python
result.all_messages()
```

## add memory to the agent

```python
result = await chat_agent.run("what is the weather in stockholm?")
result
```

```python
result.output
```

```python
result2 = await chat_agent.run("What did I ask you first?",
                              message_history=result.all_messages()
)
result2.output
```
